```python
#count the vowels in a string
string_var="Computer Science"
count=0
for x in range(len(string_var)):
  if string_var[x] in "aeiou": #this checks if alphabet of string_var is in aeiou or not
    count+=1
  else:    #it's not neccessary to write else
    pass

print(f"Vowels in {string_var} are {count}")
```

```python
#Count the upper, lower letters, digits and special characters in a string using a for loop
task_str="ComputerScienCE1237_@##$%%^^^&"
upper_count=0
lower_count=0
digit_count=0
special_count=0
for x in range(len(task_str)):
  if task_str[x].isupper():
    upper_count+=1
  elif task_str[x].islower():
    lower_count+=1
  elif task_str[x].isdigit():
    digit_count+=1
  else:
    special_count+=1

print(f"{task_str} has upper letters={upper_count}, lower letters={lower_count}, digits={digit_count} and special characters={special_count}")


```

```python
#program to enter an integer then sum the numbers, pass when 0 or ignore 0
#indexing is not possible in integer
num=int(input("Enter a number:"))
sum=0
for x in range(len(str(num))):
  if str(num)[x]=="0":   #num[x] is not valid because there is no indexing in number/integer
    pass
  else:
    sum+=int(str(num)[x])

print(sum)
```
