```python
#Different data type can also be stored in a list as well
L=['Husain', 'Rayyan', "Mukesh", '1', '2', '3']
print(L[0])
print(L[-6])
print(len(L))
print(type(L))
print(L*2)
print(L+L)
print("Husain" in L)
for i in L:
  print(i)
```

```python
#List Operations
L1=[1,2,3,4]
L2=[5,6,7,8]
print(L1*2) #Repetition
print(L1+L2)#Concatenation
print(2 in L1)#Membership
for i in L1: #Iteration
  print(i)
```

```python
fruits=['apple', 'banana', 'cherry', 'strawberry']
x=input("Enter fruit:")
if x in fruits:
  print(f"Yes {x} is in fruits")
else:
  print(f"{x} is not in fruits")
```

```python
#Changing item value in list
L=["apple", "blacberry", "mango", "apple", "1", "2", "3", "4", "a", "b", "c", "d"]

print(L)

L[1]="peach"
print(L)

L[1:3]=["banana","z"] #here 3 is exclusive, it will be bound to index 2   #At index 1 watermelon takes the place of index 1 and the value which was before the operation gets deleted, same goes with index 3 which is basically 2 beacuse 3 is exclusive
print(L)

L[1:3]=["watermelon"]
print(L)

L[1:5]=["dragonfruit"]
print(L)
```

```python
#List Functions
#Insert Function
L=["apple", "mango", "1", "2", "3"]
print(L)
L.insert(2,"apricot") #At index 2 apricot is inserted and the value before it does not gets deleted. It is moved +1
print(L)
```

```python
#List Functions
#Append Function
L=["apple", "mango", "1", "2", "3"]
print(L)
L.append("orange") #To add an item at the end of list
print(L)
```

```python
#List Functions
#Extend Function
L=["apple", "mango", "1", "2", "3"]
print(L)
L.extend(["papaya","blackberry"]) #To add 2 or more than 2 items at the end of list
print(L)
```

```python
#List Functions
#Remove Function
L=["apple", "mango", "1", "2", "3"]
print(L)
L.remove("mango") #To remove an item from the list
print(L)
```

```python
#List Functions
#Pop Function
#To remove specified index and use pop() to remove last element
L=["apple", "mango", "1", "2", "3"]
print(L)
L.pop(1) #To remove specified index and use pop() to remove last element
print(L)
L.pop()
print(L)
```

```python
#List Functions
#Del Function
#Del also remove specified index. The del keyword can also be used to delete the list
L=["apple", "mango", "1", "2", "3"]
print(L)
del(L[1])
print(L)
del(L) #The whole list gets deleted.
```

```python
#The clear method empties the list. The list still remains, but it has no content L.clear()
L=["apple", "mango", "1", "2", "3"]
print(L)
L.clear()
print(L)
```

```python
#Copy and Join List
L=["a", "apple", "c", "d", "e","f","g","orange"]
A=L.copy()
print(A)

B=list(A)
print(B)

new_list=["haha", "hehehe", "hhohoh"]
D=B + new_list
print(D)
```
