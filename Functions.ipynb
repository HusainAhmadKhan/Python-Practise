**Functions**

```python
def my_function(fname):
  print(fname + " Apple")

my_function("Red")
my_function("Yellow")
my_function("Green")
```

```python
#Different types can also be passed into a functions as parameters like string and int
def my_function(fname,num):
  print((fname + " Apple"), num)

my_function("Red",1)
my_function("Yellow",2)
my_function("Green",3)
```

```python
def my_function(fname, lname):
  print(fname, lname)

my_function("Red", "apple")
my_function("Yellow", "banana")
my_function("Green", "chilli")
```

```python
def my_function(fname, mname, lname):
  print(fname, mname, lname)

my_function("Husain", "Ahmad", "Khan")
```

```python
#Default Parameter Value
#If we call the function without argument, it uses the default value
#Passing the parameter, it overwrite the value for e.g here Karachi overwrite Pakistan
def my_function(country= "Pakistan"):
  print("I am from " +country)

my_function("Karachi")
my_function("Lahore")
my_function()
my_function("DI Khan")
```

```python
#Default arguments must come last
#Non-default parameters must come before default parameters.
#You can’t put fname="Husain" before non-default arguments.
#You can’t skip a positional argument like my_function(, "Ahmad", "Khan").
def my_function(fname="Husain", mname="", lname=""):
  print(fname, mname, lname)

my_function(fname="Hasan", mname="Ahmad", lname="Khan")
my_function("Kabir",mname="Ahmad", lname="Khan")
my_function(mname="Ahmad", lname="Khan")
my_function("Masud", "Ahmad", "Khan")
my_function("Ahmad", "Khan") #Ahmad goes to fname, Khan goes to mname, lname remains empty
my_function()
```

```python
#Returning through variables
def circle(r):
  area= 3.14*r**2
  circumference= 2*3.14*r
  return area, circumference

print(circle(5))

x=7
print(circle(x))

r=3
print(circle(r))

area, circumference= circle(r) #done by me
print(area)
print(circumference)
```

```python
#Accessing values of tuples
def circle(r):
  area= 3.14*r**2
  circumference= 2*3.14*r
  return area, circumference

result=circle(5)
print("area=", result[0],"and","circumference=", result[1])
```

```python
#Function which takes two numbers as an input and return the greater of two numbers
def greater(num1,num2):
  if num1>num2:
    return num1
  else:
    return num2

print(greater(8,7))
print(greater(7,8))
print(greater(6,6))




```
