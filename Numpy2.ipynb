```python
import numpy as np
```

```python
arr2d=np.arange(36).reshape(4,9)
print(arr2d)
```

```python
# [rows:ending, columns:ending]
arr2d[:,[5,8,0,3]]    #all rows, and dispays index 5,8,0,3 of every row
```

```python
arr2d[[0,2],[2,7]]  #[0,2]=2 , [2,7]=25   ,  you can't do arr2d[[0,2],[2,7], [2,2]] because it will be 3 dimensional, however you declared above reshape(4,9) which is 2 dimensional
```

```python
arr2d[:, ::3] #steps/alternate, every 3rd index of every row like 0,3,6
```

```python
arr2d[::3, ::3]   #every 3rd row and every 3rd column of that specific row
```

```python
arr2d[:, ::-2]  #every 2nd index of every row of reverse array  ,  minus(-) means reverse
```

```python
arr3d= np.arange(64).reshape(4,4,4)
arr3d
```

```python
print(arr3d[1][2][2])  #this can also be written as arr3d[1,2,2]
#print(arr3d[1,2,2])
```

```python
print(arr3d[3][1][3])
```

```python
print(arr3d[3])
print(arr3d[3][1:3])
```

```python
#[[52 53 54 55]  this is index 0
#[56 57 58 59]]  this is index 1
#now if you do last[1:3] 1st index which is [56 57 58 59] is displayed and 3rd index doesnt exist which is basically 2 doesnt exist it is not displayed
print(arr3d[3][1:3][1:3])
print(arr3d[3][1:3][2:3]) #the index which doesn't exist is not dispayed.
```

```python
print(arr3d[3][1:3][1:3,1:3])
```

```python
print(arr3d[3][1:3][1:3,1:5])  # since 5 index which is basically 4 doesn't exist therefore till 3rd column it gets printed which is it's last.
```

```python
print(arr3d[3][1:3][1:3,4:5]) ##the index which doesn't exist is not dispayed.
```

```python
print(arr3d)
```

```python
# [frame, rows, columns]
# [fstart:fend , rstart:rend, cstart:cend]
arr3d[3:4, 1:3, 1:3]
```

```python
print(arr3d[2, 1:5, 1:4])
print("Gap\n")
print(arr3d[1, 2, 2: ])
print("Gap\n")
print(arr3d[1:3, 1:3, 1:1])  #here 1:1 means no value
print("Gap\n")
print(arr3d[1:2, 3, 1:3])
print("Gap\n")
```

```python
#Printing different slicing for 3 dimensional array inside a function
def my_function():
  print(arr3d[3:4, 1:3, 1:3])
  print("Gap\n")
  print(arr3d[2, 1:5, 1:4])
  print("Gap\n")
  print(arr3d[1, 2, 2: ])
  print("Gap\n")
  print(arr3d[1:3, 1:3, 1:1])  #here 1:1 means no value
  print("Gap\n")
  print(arr3d[1:2, 3, 1:3])
  print("Gap\n")

my_function()
```
