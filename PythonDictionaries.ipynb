```python
#Program 5
#The keys() method will return a list of all the keys in the dictionary.
dict={
    "1": "corolla",
    "2": "alto",
    "3": "ferrari",
    "4": "altis",
    "5": "cultus",
    "6": "city",
    "7": "1993"
}

x=dict.keys()
print(x)

dict["8"]="Alto VXL"  #Add a new item to the original dictionary, and see that the keys list gets updated as well.
print(x)
```

```python
#Program 6
#The values() method will return a list of all the values in the dictionary.
dict={
    "1": "corolla",
    "2": "alto",
    "3": "ferrari",
    "4": "altis",
    "5": "cultus",
    "6": "city",
    "7": "1993"
}

a=dict.values()
print(a)

dict["year"]=2022
print(a)
```

```python
#Program 7
#The items() method will return each item in a dictionary, as tuples in a list.
dict={
    "1": "corolla",
    "2": "alto",
    "3": "ferrari",
    "4": "altis",
    "5": "cultus",
    "6": "city",
    "7": "1993"
}

a=dict.items()
print(a)

dict["year"]=2022
print(a)
```

```python
#Program 9
#Change and update
dict={
    "1": "corolla",
    "2": "alto",
    "3": "ferrari",
    "4": "altis",
    "5": "cultus",
    "6": "city",
    "7": "1993"
}

dict["4"]="apple" #You can change the value of a specific item by referring to its key name.
print(dict)

dict.update({"4": "altis"})
print(dict)


```

```python
#Program 10
#The pop() method removes the item with the specified key name
#The popitem() method removes the last inserted item (in versions before 3.7, a random item is removed instead).
#The del keyword removes the item with the specified key name. The del keyword can also delete thedictionary completely.
#The clear() method empties the dictionary.
dict={
    "1": "corolla",
    "2": "alto",
    "3": "ferrari",
    "4": "altis",
    "5": "cultus",
    "6": "city",
    "color": "green"
}
print(dict)

dict.pop("color")
print(dict)

dict.popitem()
print(dict)

del(dict["1"])
print(dict)

dict.clear()
print(dict)
```

```python
#Task 1:Generate a code which take a list of numbers and provide square of these numbers in a new list.
numbers=[1, 2, 3, 4]
squares=[]

for x in numbers:
  squares.append(x**2)

print("Orginial numbers: ",numbers)
print("Squares: ",squares)
```

```python
#Task 2:Write a code which take list of numbers as in input from user and check generate a list of numbers which is divisible by 2. Those number which are divisible
#by 2 will be saved in new list.
numbers=input("Enter numbers separated by commas:")
numbers_list=[]

for x in numbers.split(","):
  numbers_list.append(int(x))

even_numbers=[]
for x in numbers_list:
  if x % 2 == 0:
    even_numbers.append(x)

print("Original numbers: ",numbers_list)
print("Numbers divisible by 2: ",even_numbers)
```

```python
#Task 3: Write a code which take two list of numbers (l1,l2) as in input and find the common elements and save it in new list.
l1_input=input("Enter numbers for first list separated by commas: ")
l1=[]

for num in l1_input.split(","):
  l1.append(int(num))

l2_input=input("Enter numbers for second list separated by commas: ")
l2=[]

for num in l2_input.split(","):
  l2.append(int(num))

common=[]
for num in l1:
 if num in l2:
  common.append(num)

print("First List:",l1)
print("Second List:",l2)
print("Common Elements:", common)
```

```python
#Task 4 : Write a code which take an input number and provides a dictionary of
#cubic values till that number. Key is number, value is the cube as follows:
#{1:1, 2:8, 3:27, ....};
num=int(input("Enter a number: "))
cubes={}

for x in range(1,num+1):
  cubes[x]=x**3

print(cubes)
```

```python
#Task 5: Write a code which takes four inputs from user name, age, fav_sports as a
#list and fav_hobbies also as a list and print them in the following format.
#  name= ameem   fav_sport= ['cricket', ' football’]
#  age= 30   fav_hobbies= ['eating', ' playing']

name=input("Enter Name: ")

age=input("Enter Age: ")

fav_sports=input("Enter Favorite Sports separated by commas: ")

fav_hobbies=input("Enter Favorite Hobbies separated by commas: ")

fav_sports=fav_sports.split(",")

fav_hobbies=fav_hobbies.split(",")

print(f"\nName: {name}\nAge: {age}\nFavorite Sports: {fav_sports}\nFavorite Hobbies: {fav_hobbies}")
```
