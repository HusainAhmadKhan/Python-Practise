**Task 14: Broadcasting Logic**

```python
#Create a (3x3) matrix using arange().
#Create a 1D array of size 3.
#• Add them using broadcasting
#• Explain how broadcasting works

matrix_3x3 = np.arange(9).reshape(3, 3)
array_1d = np.array([10, 20, 30])
result = matrix_3x3 + array_1d

print("3x3 Matrix:\n", matrix_3x3)
print("\n1D Array:\n", array_1d)
print("\nResult of Addition:\n", result)

print("\n")
print("How Broadcasting Works?")
print("Alignment: It aligns the shapes starting from the rightmost dimension.")
print("Virtual Stretching: Since the 1D array has only one row, NumPy \"stretches\" (conceptually duplicates) that row downward until it matches the 3 rows of the matrix.")
print("Element-wise Addition: Now that the shapes are effectively both (3, 3), it adds them normally.")
```
