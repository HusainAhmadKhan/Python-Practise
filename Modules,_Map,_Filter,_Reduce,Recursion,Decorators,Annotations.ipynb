```python
#Argument parser
# Import the built-in argparse module, which is used to easily handle
# and parse command-line arguments passed to the script.
import argparse

# Create an ArgumentParser object. The 'description' parameter
# explains what the program does when a user types '--help' in the terminal.
parser = argparse.ArgumentParser(
    description = "This prints a color"
)

# Define a command-line option (-c or --color) with specific rules:
# - metavar="color": Changes the placeholder name shown in the help menu to 'color'.
# - required=True: Makes this argument mandatory; the script will throw an error if omitted.
# - choices={"red", "yellow"}: Restricts the allowed inputs so the user can *only* pick red or yellow.
# - help="..." : Provides a short description of what this option does in the help menu.
parser.addArgument("-c", "--color", metavar="color", reuqired=True, choices={"red", "yellow"}, help="The color to search for")

# Read and process the arguments provided by the user in the terminal,
# storing the results inside the 'args' object.
args = parser.parse_args()

# Access the value of the color argument using dot notation (args.color)
# and print it out to the console.
print(args.color)
```

```python
#Lamda functions
# Lambda functions are small, anonymous (unnamed) functions defined with a single expression instead of 'def'.
number = lambda num : num*2 # Defines a lambda that doubles its input; calling it without an argument causes a TypeError.
print(number(2))

multiply = lambda a,b : a*b # Defines a lambda that multiplies two numbers and prints 8 when passed 2 and 4.
print(multiply(2,4))
```

```python
#map(), filter() and reduce()
#map
# Demonstrate the built-in map() function, which applies a given function to all items in an input list
numbers = [1, 2, 7 ,8]
def double(a):
  return a*2

# double = lambda a : a*2 #this can also be done using a lambda function

# Use map() to apply the 'double' function to each element in the 'numbers' list.
# This returns a map object (an iterator), not a list directly.
result = map(double, numbers)

# or we can directly pass lambda function in map
# result = map(lambda a : a*2, numbers)

# Printing 'result' directly shows the map object's iterator type/memory location,
# because map() uses lazy evaluation.
print(result)

# Convert the map iterator into a list using list() so all values are computed and printed out.
print(list(result))

#filter
```

```python
#filter
numbers = [1, 2 , 3 , 4, 5]

def isEven(num):
  return num % 2 == 0

even_num = filter(isEven, numbers)
print(even_num)
print(list(even_num))
```

```python
from functools import reduce

expenses = [
    ("Dinner", 80),
    ("Car repair", 120)
]

# Use reduce with a lambda function to extract and add the amount (index 1) of the expenses together.
# (Note: Without an initializer, this works here because there are only two items;
# with 3+ items, the accumulated result would become an integer and cause a TypeError on subsequent steps).
total = reduce(lambda a, b: a[1] + b[1], expenses)
print(total)
```

```python
from functools import reduce

expenses = [
    ("Dinner", 80),
    ("Car repair", 120),
    ("Groceries", 50),
    ("Coffee", 10),
    ("Books", 5)
]

# The third argument (0) is the initializer.
# - On the first step, 'acc' starts at 0, and 'b' is the first tuple ("Dinner", 80).
# - Without this 0, the first iteration would add two tuples together, and the second
#   iteration would crash because 'acc' would become an integer (80) while 'b' is a tuple.

total = reduce(lambda acc, b: acc + b[1], expenses, 0) # - 'acc' (accumulator) is used to hold the running total as the loop progresses.
# - 'b' represents the current item (tuple) being looked at in the list during each step.
#0 is the initial value

print(total)  # Output: 250
```

```python
#Recursion
def factorial(n):
 if n==1:
  return 1
 return n * factorial(n-1)

print(factorial(5))
print(factorial(8))
```

```python
# Decorators: Functions that modify or enhance other functions without changing their actual code.

# Define the decorator function named 'logtime' that takes another function ('func') as input.
def logtime(func):

    # Define a wrapper function that will wrap around the original function.
    def wrapper():
        print("before")     # Code to run BEFORE the original function executes.
        val = func()        # Execute the original function and save its return value.
        print("after")      # Code to run AFTER the original function executes.
        return val          # Return the result of the original function.

    return wrapper          # Return the wrapper function itself.

# Use the @ syntax to apply the 'logtime' decorator to the 'hello' function.
# This is equivalent to writing: hello = logtime(hello)
@logtime
def hello():
    print("hello")

# Call the decorated 'hello' function.
# Because of the decorator, this will print "before", then "hello", and then "after".
hello()
```
