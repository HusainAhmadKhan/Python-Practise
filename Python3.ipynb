```python
#Enter temperaure in celsius then convert it into fahrenheit then if greater than or equal to 100 display heat warning else temperature is normal
celsius_temperature=float(input("Enter temperature in celsius:"))
fahrenheit_temperature=(celsius_temperature * (9/5)) + 32 #formula to convert celsius into fahrenheit
if fahrenheit_temperature>=100:
   print("heat warning!!!!!!!")
else:
   print("temperature is normal")
```

```python
#Calculate percentage of 6 subjects and then compare it with grades
physics_marks=float(input("Enter your physics marks out of 100:"))
maths_marks=float(input("Enter your maths marks out of 100:"))
english_marks=float(input("Enter your english marks out of 100:"))
urdu_marks=float(input("Enter your urdu marks out of 100:"))
pakistanstudies_marks=float(input("Enter your pakistan studies marks out of 50:"))
computerscience_marks=float(input("Enter your computer science marks out of 100:"))
obtained_marks=physics_marks + maths_marks + english_marks + urdu_marks + pakistanstudies_marks + computerscience_marks
percentage=(obtained_marks/550)*100
if percentage>=90:
   print("Congratulations! You secured Grade A")
elif percentage>=80 and percentage<=89:
   print("You secured Grade B")
elif percentage>=70 and percentage<=79:
   print("You secured Grade C")
else:
   print("You are not promoted. Try Again! Fail!!!!!!!!!")





```
