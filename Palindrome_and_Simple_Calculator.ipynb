```python
#Simple Calculator
num1=float(input("Enter first number: "))
num2=float(input("Enter second number: "))
operator=input("Enter operator: ")
result=0
if operator=="+":
  result=num1+num2
  print(f"{num1} + {num2} = {result}")
elif operator=="-":
  result=num1-num2
  print(f"{num1} - {num2} = {result}")
elif operator=="/":
  if num2==0:
    print("Infinityyyyyy")
  else:
   result=num1/num2
   print(f"{num1} / {num2} = {result}")
elif operator=="*":
  result=num1*num2
  print(f"{num1} * {num2} = {result}")
else:
  print("Error!!! Enter correct operator")

```

```python
#using for loop without range and and len function
word="Hello world"
count=0  #to count the length of string
for i in word:  # we can also use variable here
  count+=1
  print(i)

print(count) #displays the length of string
```

```python
#to count the number of words in a sentence
word=" Hello world! I am Husain. "
space=0
for i in word.strip():  #strip to remove extra space in start or end
  if i==" ":
    space+=1
print(f"Number of words in \"{word}\" are {space+1}")
```

```python
#PALINDROME BY SIR
word="markram"
rev_word=""
for i in word:
  rev_word = i + rev_word  #This is the logic behind palindrome
if rev_word==word:
  print("Palindrome")
else:
  print("Not a palindrome")
```
