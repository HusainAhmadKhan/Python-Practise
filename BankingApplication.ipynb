```python
#break statement
i=1
while i<6:
  print(i)
  if i==3:
    break
  i+=1
```

```python
#continue statement
i=0
while i<6:
  i+=1
  if i==3:
    continue
  print(i)
```

```python
#Challenge 3 Not completed
#Print a half and full pyramid using loop
rows=int(input("Enter rows:"))
for i in range(1,rows+1):  #right angle
   print("*"*i)

for j in range(1,rows+1):  #left angle
   print((" "*(rows-j))+"*"*j)

for x in range(1,rows+1):
   print(" "*(rows-x) + "*"*(2*x -1))
```
