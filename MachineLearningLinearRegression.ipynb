```python
#import the required libraries
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
# Creating a simple dataset
# The rows represent your individual data samples (5 samples), and the columns represent your features (1 feature).
X=np.array([[1],[2],[3],[4],[5]]) # In machine learning libraries like scikit-learn, the input features (X) must always be a 2D matrix.
y=np.array([1, 2, 3, 4, 5])
```

```python
# Fitting the Linear Regression Model
#Linear Regression is a machine learning algorithm that finds the best-fitting straight line through your data points to predict a continuous outcome(numerical value)
# It uses the relationship between an input variable (X) to predict the value of an output variable (y).
model= LinearRegression()
model.fit(X,y)
```

```python
#Making Predictions
predictions=model.predict(X)
print(predictions)
```

```python
#Visualizing the Regression Line
import matplotlib.pyplot as plt
plt.scatter(X, y, color='blue') #Scatter plot of data points
plt.plot(X, predictions, color='red') #Regression Line
plt.show()
```
