```python
import numpy as np
```

```python
#Syntax is np.zeros(size, type)  not neccessary to give a type, size should be given.
#Most answers in numpy are in float
#np.zeros gives answer in float by default
print(np.zeros(10)) #Output: Array of length 10 all 0 in  float
print(np.zeros(10, dtype="int")) #This displays in integer
print(np.zeros(10, dtype="bool")) #This displays in bool. All False
print(np.zeros(10, dtype="complex")) #This displays in complex.
print(np.zeros(10, dtype="str")) #This displays in string.
```

```python
print(np.zeros((4,5))) #This is a tuple passed here. tuple is (4,5)
```

```python
print(np.zeros((4,4,4), dtype="int")) #This is a three dimensional array in integer
```

```python
print(np.ones(10)) #Output: Array of length 10 all 0 in  float
print(np.ones(10, dtype="int")) #This displays in integer
print(np.ones(10, dtype="bool")) #This displays in bool. All True
print(np.ones(10, dtype="complex")) #This displays in complex.
print(np.ones(10, dtype="str")) #This displays in string.
```

```python
#Matrix Multiplication
# Column of first == Rows of second (Checking Operation)
# Row of first * Column of second (Multiply Operation)
a=np.arange(10).reshape(5,2)
b=np.arange(10).reshape(2,5)
print(a@b) # @ is used for matrix multiplication
```
